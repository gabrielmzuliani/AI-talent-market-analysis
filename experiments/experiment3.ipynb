Scenario 3 - Smaller companies are inherently more flexiblt than large companies?

In this scenario, I want to verify whether smaller companies are more inherent with flexible work modes than large companies. To test this assumption, I am going to consider both remote and hybrid work modes as flexible and compare them with on-site mode. I decided to follow this rule because the spread of hybrid data is strictly less than the others, and the hybrid work mode doesn't cease to be a flexible work mode like remote work is.

* Null Hypothesis (H0): Company size and work mode are completely independent variables.
* Alternative Hypóthesis (Ha): There is a statistically significant association between a company's size and the work modes it offers.

```python
# libraries
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

plt.style.use('ggplot')

from matplotlib.pyplot import figure
from matplotlib.lines import Line2D

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12, 8)
```

1 - Data structure

```python
# reading the data
df = pd.read_csv(
    r'../data/ai_jobs_salaries_clean.csv'
)

print(f'| Rows: {df.shape[0]} | Columns: {df.shape[1]} |')
df.head()
```

```python
# combining hybrid and remote work modes within a single category
df['work_mode_lb'] = np.where(
    (df['work_mode'] == 'Remote') | (df['work_mode'] == 'Hybrid'), 'Flexible', 'Non-Flexible'
)

count_wm = df['work_mode_lb'].value_counts()
perc_wm = round(df['work_mode_lb'].value_counts(normalize = True) * 100, 1).astype(str) + ' %'

print('-' * 37)
print('| Work Mode    | Count | Percentage |')

for i in count_wm.index:
    print(f'| {i:<12} | {count_wm[i]:<5} | {perc_wm[i]:<10} |')

print('-' * 37)
```

```python
# assigning a label for the company size 
df['company_size_lb'] = df['company_size'].map(
    {
        'S': 'Small',
        'M': 'Medium',
        'L': 'Large'
    }
)

print(df['company_size_lb'].value_counts())
```

```python
# saving the dataset
df.to_csv(r'../data/df_compsize_wkmode_corr.csv')
```

4 - Applying the test

**Experiment 1:** Are different company sizes more related to flexibility work culture?

**Variables:** 
* Company Size (Small, Medium, Large); 
* Work Mode (Remote and On-site)

**Test Parameters:**
* **Null Hypothesis (H0):** Company size and work mode are completely independent variables
* **Alternative Hypothesis (H1):** There is a statiscally significant association between company sizes and work mode
* **Alpha:** 0.05

```python
# creating the contingency table
contingency_table = pd.crosstab(
    df['company_size_lb'],
    df['work_mode_lb']
)

print(contingency_table)
```

```python
# performing the chi-square test
stat, p_value, dof, expect = stats.chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {stat:.2f}")
print(f"P-value: {p_value:.3e}\n")
```

```python
# evaluate the hypothesis
alpha = 0.05

if p_value < alpha:
    print("Reject H0: There is a statistically significant association between company size and work mode.")
else:
    print("Fail to Reject H0: Company size and work mode are strictly independent.")
```
